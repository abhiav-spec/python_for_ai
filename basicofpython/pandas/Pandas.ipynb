```python
import numpy as np
import pandas as pd

# 5 rows and 6 columns requires exactly 30 data points (0 to 29)
df = pd.DataFrame(
    data=np.arange(0, 30).reshape(5, 6),
    index=["Emp_01", "Emp_02", "Emp_03", "Emp_04", "Emp_05"],
    columns=[
        "Age",
        "Experience_Years",
        "Projects_Completed",
        "Performance_Score",
        "Monthly_Salary_K",
        "Training_Hours",
    ],
)

print(df)
```
